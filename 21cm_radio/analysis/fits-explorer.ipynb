```python
from astropy.io import fits
import numpy as np
import matplotlib.pyplot as plt

path = "/Users/elliottchalcraft/Documents/2026-27/8.13/jlab/21cm_radio/data/raw/10-19-2026/sun/SRT_SPEC_SAVE-2026_10_19_14_05_28.fits"

hdul = fits.open(path)
hdul[0].header
```

```python
data = hdul[12].data
header = hdul[0].header

channel0 = data[0, 0, :, 0]
channel1 = data[1, 1, :, 0]

freqs = header["CRVAL2"] + (np.arange(512) - header["CRPIX2"]) * header["CDELT2"]
# print(freqs)

power = channel0**2 + channel1**2
# print(power)

plt.figure()
plt.plot(freqs,power)
plt.show()

# hdul[0].header['BUNIT']
```
